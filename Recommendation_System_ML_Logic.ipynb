Making Data Readable to get Key Idea

```python
import ast
def convert(obj):
    L=[]
    for i in ast.literal_eval(obj):
      L.append(i['name'])
      if len(L) >= 3: # Limit to first 3 names
          break
    return L
```

```python
movies['genres']=movies['genres'].apply(convert)
```

```python
movies.head()
```

```python
movies['keywords']=movies['keywords'].apply(convert)
```

```python
movies.head()
```

```python
movies['cast']=movies['cast'].apply(convert)
```

```python
movies.head()
```

```python
import ast
def fetch_director(obj):
    L=[]
    try:
        for i in ast.literal_eval(obj):
            if i['job']=='Director':
                L.append(i['name'])
                break
    except (ValueError, SyntaxError):
        pass
    return L
```

```python
movies['crew']=movies['crew'].apply(fetch_director)
```

```python
movies.head()
```

```python
movies['overview']=movies['overview'].apply(lambda x:x.split())
```

```python
movies.head()
```

meking a complete word to avoide the ambuguities like following:

Sam Worthington = SamWorthington

Sam Mendes =SamMendes

```python
movies['genres']=movies['genres'].apply(lambda x:[i.replace(" ","") for i in x])
movies['keywords']=movies['keywords'].apply(lambda x:[i.replace(" ","") for i in x])
movies['cast']=movies['cast'].apply(lambda x:[i.replace(" ","") for i in x])
movies['crew']=movies['crew'].apply(lambda x:[i.replace(" ","") for i in x])
```

```python
movies.head()
```

Creting New Column 'TAG' Which Made by Combinatons of overview,genres,keywords,creW

```python
movies['tags']=movies['overview']+movies['genres']+movies['keywords']+movies['cast']+movies['crew']
```

```python
movies.head()
```

```python
new_df = movies[['id','title','tags']]
```

```python
new_df
```

```python
new_df['tags']=new_df['tags'].apply(lambda x:" ".join(x))
```

```python
new_df['tags'][0]
```

```python
new_df['tags']=new_df['tags'].apply(lambda x:x.lower())
```

```python
new_df.head()
```

NOW PERFORMING TEXT VECTORIZING
BY USING BAG OF WORDS


```python
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer(max_features=5000,stop_words='english')
```

```python
vectors=cv.fit_transform(new_df['tags']).toarray()
```

```python
vectors
```

```python
vectors[0]
```

```python
cv.get_feature_names_out()
```

Apply Stemming to Neglate Repetitions like [loved,loving,love]=[love]

```python
pip install  nltk
```

```python
import nltk
```

```python
from nltk.stem.porter import PorterStemmer
ps=PorterStemmer()
```

```python
def stem(text):
  y=[]
  for i in text.split():
    y.append(ps.stem(i))
  return " ".join(y)
```

```python
ps.stem('loved')
```

```python
new_df['tags']=new_df['tags'].apply(stem)
```

```python
vectors
```

```python
vectors[0]
```

```python
cv.get_feature_names_out()
```

Similarity Computation: cosine distance

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity=cosine_similarity(vectors)
```

```python
similarity.shape
```

```python
similarity[1]
```

Recommendation Function

```python
def recommend(movie):
  movie_index=new_df[new_df['title']==movie].index[0]
  distances=similarity[movie_index]
  movies_list=sorted(list(enumerate(distances)),reverse=True,key=lambda x:x[1])[1:6]
  for i in movies_list:
    print(new_df.iloc[i[0]].title)
  return
```

To Support Lowercase as well as Uppercase

```python
def recommend(movie):
    movie = movie.lower()
    new_df['title_lower'] = new_df['title'].str.lower()

    if movie not in new_df['title_lower'].values:
        print("Movie not found ❌")
        return

    index = new_df[new_df['title_lower'] == movie].index[0]
    distances = similarity[index]
    movies_list = sorted(list(enumerate(distances)), reverse=True, key=lambda x:x[1])[1:6]

    print("Recommended movies:")
    for i in movies_list:
        print(new_df.iloc[i[0]].title)
```

Testing the System

```python
recommend('Avatar')
```
